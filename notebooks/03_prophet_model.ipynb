```python
import pandas as pd
from prophet import Prophet
from sklearn.metrics import mean_absolute_percentage_error, root_mean_squared_error, mean_absolute_error
```

```python
print("Loading data for Prophet...")
df = pd.read_csv('data/bike_sharing.csv')
df['date'] = pd.to_datetime(df['date'])
```

```python
# Prophet requires 'ds' and 'y' columns
pdf = df[['date', 'demand']].rename(columns={'date': 'ds', 'demand': 'y'})
```

```python
train_size = int(len(pdf) * 0.8)
train, test = pdf.iloc[:train_size], pdf.iloc[train_size:]
```

```python
print("Fitting Prophet...")
model = Prophet(yearly_seasonality=True, weekly_seasonality=True, daily_seasonality=True)
```

```python
# Add holidays if possible
holidays = df[df['is_holiday'] == 1][['date']].rename(columns={'date': 'ds'})
holidays['holiday'] = 'public_holiday'
model = Prophet(yearly_seasonality=True, weekly_seasonality=True, daily_seasonality=True, holidays=holidays)
```

```python
model.fit(train)
```

```python
future = test[['ds']]
forecast = model.predict(future)
```

```python
preds = forecast['yhat'].values
y_true = test['y'].values
```

```python
prophet_mape = mean_absolute_percentage_error(y_true, preds)
prophet_rmse = root_mean_squared_error(y_true, preds)
prophet_mae = mean_absolute_error(y_true, preds)
```

```python
print(f"Prophet -> MAPE: {prophet_mape:.4f}, RMSE: {prophet_rmse:.4f}, MAE: {prophet_mae:.4f}")
```
